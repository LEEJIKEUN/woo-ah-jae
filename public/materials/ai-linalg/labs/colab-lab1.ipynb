## 0. 준비 — NumPy 불러오기

`numpy`는 파이썬에서 벡터·행렬을 다루는 도구 상자입니다.
관례적으로 `np`라는 짧은 이름을 붙여 씁니다.

```python
import numpy as np

print(np.__version__)
```

## 1. 벡터 만들기 (Vector)

숫자를 순서대로 나열한 것이 벡터입니다.
`shape`는 그 벡터의 **크기**를 알려줍니다.

```python
# 벡터 만들기
c = np.array([1, 2, 3])

print(c)
print(c.shape)   # (3,) → 숫자 3개짜리
```

```python
# 특정 성분 꺼내기 — 번호는 0부터 시작!
print(c[0])   # 첫 번째
print(c[2])   # 세 번째
```

> ⚠️ **주의** — 수학에서는 첫 번째를 1번이라 부르지만, 파이썬은 **0번**부터 셉니다.

## 2. 행벡터 (Row Vector)

대괄호를 한 겹 더 씌우면 **1행 3열 행렬**, 즉 행벡터가 됩니다.
`shape`가 `(3,)`에서 `(1, 3)`으로 바뀌는 것을 확인하세요.

```python
# 행벡터 (1행 3열)
r = np.array([[1, 2, 3]])

print(r)
print(r.shape)   # (1, 3)
```

```python
# 2차원이므로 [행, 열] 두 개의 번호로 꺼냅니다
print(r[0, 1])   # 0행 1열 → 2
```

## 3. 자주 쓰는 행렬 빠르게 만들기

매번 숫자를 다 적을 필요는 없습니다.

```python
# 모두 0인 행렬
a = np.zeros((2, 2))
print(a)

# 모두 1인 행렬
b = np.ones((2, 2))
print(b)
```

```python
# 같은 값으로 채우기
c = np.full((2, 2), 7)
print(c)

# 무작위 값 채우기 (실행할 때마다 달라집니다)
d = np.random.random((2, 2))
print(d)
```

## 4. 행렬 만들기와 전치 (Matrix & Transpose)

안쪽 대괄호 하나가 **한 줄(행)** 입니다.

```python
# 3행 2열 행렬
A = np.array([[1, 2],
              [3, 4],
              [5, 6]])

print(A)
print(A.shape)   # (3, 2)
```

```python
# 전치 — 행과 열을 뒤집습니다
print(A.T)
print(A.T.shape)   # (2, 3) ← 크기도 뒤집힘!
```

## 5. 행렬 곱셈 (Matrix Multiplication)

파이썬에서 행렬 곱은 **`@`** 기호를 씁니다.
`np.dot(A, B)`, `A.dot(B)` 도 같은 뜻이지만 `@`가 가장 읽기 쉽습니다.

```python
B = np.array([[11, 12, 13, 14],
              [15, 16, 17, 18]])

print(A.shape, B.shape)   # (3, 2) (2, 4) → 가운데 2가 일치!

print(A @ B)
```

### 크기가 안 맞으면?

아래 셀은 **일부러 오류를 내는** 셀입니다. 빨간 에러 메시지가 나오는 게 정상입니다.

```python
# B는 (2, 4), A는 (3, 2) → 4 ≠ 3 이므로 곱할 수 없습니다
print(B @ A)
```

> 에러 메시지의 마지막 줄을 읽어 보세요.
> `shapes (2,4) and (3,2) not aligned` — **가운데 두 숫자가 안 맞는다**는 뜻입니다.

### `*` 와 `@` 는 완전히 다릅니다

```python
P = np.array([[1, 2],
              [3, 4]])
Q = np.array([[5, 6],
              [7, 8]])

print(P * Q)    # 같은 자리끼리 곱함 (성분별 곱)
print()
print(P @ Q)    # 진짜 행렬 곱셈
```

## 6. 교환법칙은 성립하지 않는다

수업에서 배운 `AB ≠ BA` 를 직접 확인해 봅시다.

```python
print(P @ Q)
print()
print(Q @ P)
print()
print("두 결과가 같은가?", np.array_equal(P @ Q, Q @ P))
```

### 오늘 배운 명령어 정리

| 코드 | 뜻 |
|---|---|
| `np.array([...])` | 벡터·행렬 만들기 |
| `.shape` | 크기 확인 (행, 열) |
| `.T` | 전치 |
| `A @ B` | 행렬 곱셈 |
| `A * B` | 성분별 곱 (행렬 곱 아님!) |
| `np.zeros`, `np.ones`, `np.full` | 특정 값으로 채운 행렬 |
| `np.random.random` | 무작위 행렬 |
