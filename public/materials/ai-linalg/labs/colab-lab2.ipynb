## 0. 준비

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)   # 소수점 4자리까지만 보기
```

## 1. 문제를 행렬로 옮기기

| 사람 | 몸무게 | 키 | 흡연 | 수명 |
|---|---|---|---|---|
| 1 | 60 kg | 5.5 ft | 1 | 66 |
| 2 | 65 kg | 5.0 ft | 0 | 74 |
| 3 | 55 kg | 6.0 ft | 1 | 78 |

$$A\mathbf{x} = \mathbf{b}$$

```python
A = np.array([[60, 5.5, 1],
              [65, 5.0, 0],
              [55, 6.0, 1]])

b = np.array([66, 74, 78])

print(A)
print(A.shape, b.shape)
```

## 2. 항등행렬 (Identity Matrix)

대각선만 1인 정사각행렬. 숫자의 **1**과 같은 역할을 합니다.

```python
I = np.eye(3)
print(I)

# 어떤 벡터에 곱해도 그대로
print(I @ b)
```

## 3. 역행렬 (Inverse Matrix)

수업에서는 2×2 공식만 배웠지만, 3×3부터는 공식이 없습니다. 컴퓨터에게 맡깁니다.

```python
from numpy.linalg import inv, det, solve

A_inv = inv(A)
print(A_inv)
```

### 정말 역행렬일까? — 곱해서 확인

$A^{-1}A = AA^{-1} = I_n$ 이어야 합니다.

```python
print(A @ A_inv)
```

> ⚠️ `-0.` 이나 `0.0000001` 같은 값이 섞여 나올 수 있습니다.
> 컴퓨터가 소수를 다룰 때 생기는 아주 작은 오차이며, **사실상 0**입니다.

### 실습 1 복습 — `*` 를 쓰면 안 됩니다

```python
print(A * A_inv)   # 성분별 곱 — 항등행렬이 나오지 않습니다!
```

## 4. 드디어 해 구하기

$$\mathbf{x} = A^{-1}\mathbf{b}$$

```python
x = A_inv @ b
print(x)
```

**수업에서 손으로 구한 답과 같은가요?**

$$(\text{수명}) = -0.4 \times (\text{몸무게}) + 20 \times (\text{키}) - 20 \times (\text{흡연})$$

### 더 좋은 방법 — `solve`

역행렬을 따로 구하지 않고 한 번에 풉니다. **더 빠르고 더 정확합니다.**

```python
x2 = solve(A, b)
print(x2)

print("두 방법의 결과가 같은가?", np.allclose(x, x2))
```

> 💡 실무에서는 `inv()` 보다 **`solve()`** 를 씁니다.
> 역행렬을 거치면 계산이 한 단계 더 늘어나 오차가 쌓이기 때문입니다.
>
> `np.allclose` 는 아주 작은 오차를 무시하고 "거의 같은가"를 판단합니다.

## 5. 검산 — 원래 식에 넣어 보기

```python
print(A @ x)      # b 와 같아야 함
print(b)
```

## 6. 행렬식 (Determinant)

$\det A \neq 0$ 이면 역행렬이 존재합니다.

```python
print(det(A))
```

### 역행렬이 없는 행렬은?

수업에서 본 행렬입니다. 2행이 1행의 2배라 정보가 겹칩니다.

$$S = \begin{bmatrix} 1 & 2 \\ 2 & 4 \end{bmatrix}$$

```python
S = np.array([[1, 2],
              [2, 4]])

print(det(S))
```

아래 셀은 **일부러 오류를 내는** 셀입니다.

```python
print(inv(S))
```

> `LinAlgError: Singular matrix`
>
> **singular(특이)** 란 역행렬이 없다는 뜻입니다. 컴퓨터도 못 만듭니다.

### 오늘 배운 명령어

| 코드 | 뜻 |
|---|---|
| `np.eye(n)` | n × n 항등행렬 |
| `inv(A)` | 역행렬 |
| `det(A)` | 행렬식. 0이면 역행렬 없음 |
| `solve(A, b)` | `Ax = b` 를 한 번에 풀기 **(권장)** |
| `np.allclose(a, b)` | 오차를 감안해 "거의 같은가" 판단 |
| `np.set_printoptions(...)` | 출력 자릿수 조절 |
